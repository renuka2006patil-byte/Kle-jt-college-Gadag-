```python
# Simple Linear Regression
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics  import r2_score
import matplotlib.pyplot as plt

df=pd.DataFrame({"Hours":[1,2,3,4,5],"Marks":[20,40,50,65,80]})


X=df[['Hours']]
Y=df['Marks']
X_train,X_test,Y_train,Y_test=train_test_split(X,Y, test_size=0.2)

model = LinearRegression()

model.fit(X_train,Y_train)

prediction = model.predict ([[7]])
plt. scatter (X, Y)

plt.plot(X, model.predict (X), color='red')

plt.title("Simple LinearRegression")

plt. xlabel("Study_hours")

plt. ylabel ("Marks")
plt.grid()

plt.legend()

plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score

#2)crete Datasets

data={'House(sqrt)':[100,200,300,400,500,600,700,800],
     'Price(Lakhs)':[25,30,35,40,45,50,55,60]
      }
df=pd.DataFrame(data)

#3)find out input and output variables
X=df[['House(sqrt)']]
Y=df['Price(Lakhs)']

#4)Train-Test model
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3)


#5)object creation of Algorithms
model=LinearRegression()

#6)call fit()function of input and output variables
model.fit(X_train,Y_train)

#7)Prediction new value
print("Prediction of new value is :",model.predict([[1200]]))

#8) calculate Metrics
MSE=mean_squared_error(Y_test,X_test)
print("Mean_squared_error value is",mean_squared_error(Y_test,X_test))
print("R2_Score value is :",r2_score(Y_test,X_test))
plt.figure(figsize=(10,6))
plt.scatter(X,Y,color="#800020",marker='*',edgecolor="#D10363",alpha=0.9,linewidth=0.4)
plt.plot(X,model.predict(X),label="Regression Line",color="#5DD3B6",linestyle='--',linewidth=2)
plt.title("Linear Regression")
plt.xlabel("Hours(SQRT)")
plt.ylabel("Price(LAKHS)")
plt.grid(linestyle="--")
plt.legend()
plt.show()
```
